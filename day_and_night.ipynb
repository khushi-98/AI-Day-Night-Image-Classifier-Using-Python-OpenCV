```python
import cv2
import gradio as gr
```

```python
def day_night_classifier(image):
  gray=cv2.cvtColor(image,cv2.COLOR_RGB2GRAY)
  brightness=gray.mean()

  if brightness>100:
    return "DAY"
  else:
    return "NIGHT"


```

```python
gr.Interface(
    fn=day_night_classifier,
    inputs=gr.Image(type="numpy"),
    outputs="text",
    title="AI Day/ Night Classifier",
    description="Classifies an image as day or night based on brightness."
).launch()
```
